# Variance and standard deviation

The code from the "Pros" section of the guide: https://franckbinde.pages.dev/en/fiches/variance/

Press the play button of each cell, in order.

```python
import pandas as pd
import numpy as np

trips = pd.DataFrame({
    "main_road": [29, 31, 30, 28, 32, 30, 29, 31, 30, 30],
    "shortcut":  [15, 50, 22, 41, 18, 35, 25, 45, 20, 29],
})
print(trips.mean())
print(trips.std().round(2))
print(np.std(trips["shortcut"]).round(2))
print((trips.mean() + 2 * trips.std()).round(1))
```
