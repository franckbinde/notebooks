```python
from sentence_transformers import SentenceTransformer, util

modele = SentenceTransformer("paraphrase-multilingual-MiniLM-L12-v2")
phrases = ["Je cherche un appartement à louer",
           "Je voudrais louer un logement",
           "La bourse a fortement monté"]
vecteurs = modele.encode(phrases)
print(vecteurs.shape)
print(util.cos_sim(vecteurs[0], vecteurs[1]))
print(util.cos_sim(vecteurs[0], vecteurs[2]))
```
