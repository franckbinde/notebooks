# Variance et écart-type

Le code de la section « Chez les pros » de la fiche : https://franckbinde.pages.dev/fiches/variance/

Appuie sur le bouton de lecture de chaque cellule, dans l'ordre.

```python
import pandas as pd
import numpy as np

trajets = pd.DataFrame({
    "grande_route": [29, 31, 30, 28, 32, 30, 29, 31, 30, 30],
    "raccourci":    [15, 50, 22, 41, 18, 35, 25, 45, 20, 29],
})
print(trajets.mean())
print(trajets.std().round(2))
print(np.std(trajets["raccourci"]).round(2))
print((trajets.mean() + 2 * trajets.std()).round(1))
```
